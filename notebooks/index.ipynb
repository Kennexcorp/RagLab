## Initialize RAG System

````python
# Initialize the RAG system
rag = RAGSystem(log_level=LOG_LEVEL)

# Display system configuration
print(Config.display_config())
````

## Data Ingestion

Load sample dashboard data into the vector store.

````python
# Ingest sample data
data_file = "sample_data.json"

num_chunks = rag.ingest_data(source=data_file, source_type="json")

print(f"\n✓ Successfully ingested {num_chunks} chunks from {data_file}")
````

````python
# Get system statistics
stats = rag.get_stats()

print("=== System Statistics ===")
print(f"Total Documents: {stats['vector_store']['document_count']}")
print(f"Collection: {stats['vector_store']['collection_name']}")
print(f"\nLLM Provider: {stats['config']['llm_provider']}")
print(f"LLM Model: {stats['config']['llm_model']}")
print(f"Embedding Model: {stats['config']['embedding_model']}")
print(f"Chunk Size: {stats['config']['chunk_size']} tokens")
print(f"Top K Results: {stats['config']['top_k']}")
````

## Interactive Query

Try your own questions!

````python
# Ask your own question
your_question = "What are the top performing areas of the business?"  # Change this!

response = rag.query(your_question, top_k=5)

print(f"Question: {response['question']}")
print(f"\nAnswer:\n{response.get('answer_with_citations', response['answer'])}")

# Show source details
if response.get("sources"):
    print("\n--- Source Details ---")
    for i, source in enumerate(response["sources"][:3], 1):
        print(f"\nSource {i}:")
        print(f"Text: {source['text'][:150]}...")
        print(f"Metadata: {source['metadata']}")
        print(f"Similarity: {source.get('similarity_score', 'N/A')}")
````

## Performance Analysis

````python
# Analyze query performance
test_queries = [
    "What is our revenue?",
    "How many users do we have?",
    "What is our customer satisfaction score?",
    "How is marketing performing?",
    "What are the infrastructure metrics?",
]

print("=== Performance Analysis ===")
for query in test_queries:
    response = rag.query(query, top_k=3)
    perf = response.get("performance", {})

    retrieval_time = perf.get("retrieval", {}).get("duration", 0)
    generation_time = perf.get("generation", {}).get("duration", 0)
    total_time = perf.get("query_processing", {}).get("duration", 0)

    print(f"\nQuery: {query}")
    print(f"  Retrieval: {retrieval_time:.3f}s")
    print(f"  Generation: {generation_time:.3f}s")
    print(f"  Total: {total_time:.3f}s")
    print(f"  Sources: {response.get('num_sources', 0)}")
````

## Advanced: Direct Component Access

You can also access individual components for more control.

````python
# Access retriever directly

retriever = rag.retriever

# Retrieve without generation
query = "revenue metrics"
results = retriever.retrieve(query, top_k=3)

print(f"Retrieved {len(results)} documents for: '{query}'\n")
for i, result in enumerate(results, 1):
    print(f"Result {i}:")
    print(f"  Text: {result['text'][:100]}...")
    print(f"  Similarity: {result.get('similarity_score', 'N/A')}")
    print()
````
